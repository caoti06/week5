File nằm ở data/Social_Network_Ads.csv.

Mỗi dòng là một khách. Age và EstimatedSalary là feature. Purchased là label: 0 là không mua, 1 là mua. 



## Step 1. Load the data


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

df = pd.read_csv("D:/w5/w5/Linear Regression & Logistic Regession/Practice/Social_Network_Ads.csv")
df.head()
```

## Step 2. Split features, label, then train and test sets

Hint:

- features are the two columns Age and EstimatedSalary
- label is Purchased
- use train_test_split with test_size = 0.2 and random_state = 0

80% of the rows are for training. 20% are for testing later.


```python
from sklearn.model_selection import train_test_split

# Features and label
X = df[["Age", "EstimatedSalary"]]
y = df["Purchased"]

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=0
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

## Step 3. Train the model

Hint: LogisticRegression(max_iter = 100), then fit on x_train and y_train.

fit loops inside the library, at most 100 rounds, and it can stop early. Print model.n_iter_[0] to see how many rounds it ran.

Print three numbers:

- intercept_ is the bias
- coef_[0][0] is the weight of Age
- coef_[0][1] is the weight of EstimatedSalary

z = intercept_ + weight_age * Age + weight_salary * EstimatedSalary


```python
from sklearn.linear_model import LogisticRegression

# Create and train the model
model = LogisticRegression(max_iter=100)
model.fit(X_train, y_train)

# Get bias and weights
intercept = model.intercept_[0]
weight_age = model.coef_[0][0]
weight_salary = model.coef_[0][1]

print("Number of iterations =", model.n_iter_[0])
print("Intercept =", intercept)
print("Weight of Age =", weight_age)
print("Weight of EstimatedSalary =", weight_salary)
```
